```python
# ============================================================
# Library Imports
# ============================================================
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import OrdinalEncoder
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from joblib import dump
```

```python
# ============================================================
# Load Datasets (Data Retrieval)
# ============================================================

# Dataset paths (Kaggle)
DISEASE_DATASET_PATH    = '/Users/abhishek/Documents/Abhishek/first_doctor/backend/data/DiseaseAndSymptoms.csv'
PRECAUTION_DATASET_PATH = '/Users/abhishek/Documents/Abhishek/first_doctor/backend/data/Disease precaution.csv'

# Load disease-symptom dataset
disease_df = pd.read_csv(DISEASE_DATASET_PATH)
print(f'✅ Disease dataset loaded: {disease_df.shape[0]} rows, {disease_df.shape[1]} columns')
print(f'📋 Unique diseases: {disease_df.iloc[:,0].nunique()}')
disease_df.head(3)
```

```python
# ============================================================
# Train KNN Classifier (Data Processing)
# ============================================================

# Extract features and labels
known_symptoms = list(disease_df.columns[1:])
X = disease_df.iloc[:, 1:].values
y = disease_df.iloc[:, 0].values

# Encode labels
ce = OrdinalEncoder()
y_encoded = ce.fit_transform(y.reshape(-1, 1)).ravel()

# Train/test split for evaluation
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Train KNN model
neigh = KNeighborsClassifier(n_neighbors=3)
neigh.fit(X_train, y_train)

# Evaluate model
y_pred = neigh.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'✅ KNN Classifier trained successfully!')
print(f'🎯 Model Accuracy on Test Set: {accuracy * 100:.2f}%')
print(f'📊 Total symptoms (features): {len(known_symptoms)}')

# Load precautions dataset
precautions_df = pd.read_csv(PRECAUTION_DATASET_PATH)
print(f'✅ Precautions dataset loaded: {precautions_df.shape[0]} diseases covered')

print('model saved successfully!')
dump(neigh, "save-models/disease_predictor(KNN).joblib")

print('label encoder saved successfully!')
dump(ce, "save-models/label_encoder(KNN).joblib")
```
